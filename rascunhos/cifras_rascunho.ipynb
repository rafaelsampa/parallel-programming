```python


from cryptography.hazmat.primitives.ciphers import Cipher, algorithms, modes
import os
```

```python


# AES-128 modo ECB


key = os.urandom(16) # 128 bits
mensagem_para_cifrar = b"Rafael Sampaio e Silva".ljust(48, b" ")

encriptar = Cipher(algorithms.AES(key), modes.ECB()).encryptor()
texto_cifrado  = encriptar.update(mensagem_para_cifrar) + encriptar.finalize()

print("cifrado: ", texto_cifrado.hex())

```

```python

# AES-128 modo CTR

# ciphertext = texto cifrado
# plaintext = texto decifrado


key = os.urandom(16)
# nonce = os.urandom(16)
# ou
# nonce = 1 # um numero qualquer, mas deveria ser difente para cada vez q for cifra algo. Nesse caso nao funciona pq tem que ser 16 bytes
nonce = (1).to_bytes(16)
# → b'\x00\x00\x00\x00\x00\x00\x00\x00\x00\x00\x00\x00\x00\x00\x00\x01'

mensagem_para_cifrar = b"Rafael Sampaio e Silva" # qualquer tamanho

encriptar = Cipher(algorithms.AES(key), modes.CTR(nonce)).encryptor()
ciphertext = encriptar.update(mensagem_para_cifrar) + encriptar.finalize()

decriptar = Cipher(algorithms.AES(key), modes.CTR(nonce)).decryptor()
plaintext = decriptar.update(ciphertext) + decriptar.finalize()


print("cifrado: ", ciphertext.hex())
print("decifrado: ", plaintext.decode())

```

```python

# ChaCha20 com bibli

key = os.urandom(32)
nonce = os.urandom(16)
mensagem_para_cifrar = b"Rafael Sampaio e Silva"

encriptar = Cipher(algorithms.ChaCha20(key, nonce), mode = None).encryptor()
ciphertext  = encriptar.update(mensagem_para_cifrar) + encriptar.finalize()

decriptar = Cipher(algorithms.ChaCha20(key, nonce), mode = None).decryptor()
plaintext = decriptar.update(ciphertext) + decriptar.finalize()

print("cifrado: ", ciphertext.hex())
print("decifrado:", plaintext.decode())


```
